```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
# Create sample education dataset
data = {
    "Year": [2018, 2019, 2020, 2021, 2022],
    "Enrollment": [520000, 530000, 515000, 540000, 560000],
    "Graduation_Rate": [78, 79, 77, 80, 82],
    "Public_Spending_Million_SAR": [180, 185, 190, 195, 200]
}

df = pd.DataFrame(data)
df
```

```python
# Data overview
df.info()
df.describe()
```

# Visualization: Enrollment trend over time


```python
# Enrollment trend
plt.figure()
plt.plot(df["Year"], df["Enrollment"], marker="o")
plt.title("Student Enrollment Over Years")
plt.xlabel("Year")
plt.ylabel("Enrollment")
plt.show()
```

```python
# Graduation rate trend
plt.figure()
plt.plot(df["Year"], df["Graduation_Rate"], marker="s")
plt.title("Graduation Rate Over Years")
plt.xlabel("Year")
plt.ylabel("Graduation Rate (%)")
plt.show()
```
